```python
x="Mridul"

def fn():
    x="Tiwari"

fn()
print(x)

# to edit var inside fn it should be declared global
def fc():
    global x
    x="Tiwari"

fc()
print(x)
```

```python
# Built in data types
x="Mridul"
print(x,type(x))
x=10
print(x,type(x))
x=10.5
print(x,type(x))
x=True
print(x,type(x))
x=["Mridul","Tiwari"]#list
print(x,type(x))
x=("Mridul","Tiwari")#tuple
print(x,type(x))
x={"Mridul","Tiwari","Tiwari"}#set
print(x,type(x))
x={"firstName":"Mridul", "lastName":"Tiwari"}#dic
print(x,type(x))
```
