```python
### Python Lists
fruits=["apple","mango","banana"] ### ordered, changeable, allow duplicate
print(fruits)
print(len(fruits))
list=["Mridul",23,True] ### a single list can store any data type
print(list)
print(type(list))
print(list[0])
print(list[0:2])
print(list[:-1])
if "Mridul" in list:
    print("found")
list[1]=25
print(list)
list[2]="Tiwari"
print(list)
list.insert(3,"Digit")
print(list)
list.append("Insurance")
print(list)
list.extend(fruits)
print(list)
list.remove(25)
print(list)
list.pop()
print(list)
del list[4] ### usefull when to remove on a particular index
print(list)
fruits.clear()
print(fruits)
for x in list:
    print(x, end=",")
print()
for i in range(len(list)):
    print(list[i], end=" ")
print()
i=0
while i<len(list):
    print(list[i], end=" ")
    i+=1
```
